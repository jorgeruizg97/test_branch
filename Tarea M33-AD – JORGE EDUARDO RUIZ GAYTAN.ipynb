## Análisis de crecimiento - Últimos 4 años

El dataset contiene registros para 2017 y 2020, pero estos años cuentan
únicamente con 3 y 1 videojuegos respectivamente, por lo que no son
representativos para realizar un análisis de crecimiento.

Por esta razón, se utilizarán los últimos cuatro años con un volumen
significativo de registros: 2013, 2014, 2015 y 2016.

```python
ultimos_4_anios = [2013, 2014, 2015, 2016]

df_ultimos4 = df_limpio[
    df_limpio["Year"].isin(ultimos_4_anios)
].copy()

df_ultimos4["Year"].value_counts().sort_index()
```

```python
crecimiento_genero = (
    df_ultimos4
    .groupby(["Year", "Genre"])["Global_Sales"]
    .sum()
    .unstack(fill_value=0)
)

crecimiento_genero
```

```python
cambio_genero = (
    crecimiento_genero.loc[2016] -
    crecimiento_genero.loc[2013]
).sort_values(ascending=False)

cambio_genero
```

```python
print("Género con mayor crecimiento:")
print(cambio_genero.head(1))

print("\nGénero con menor crecimiento:")
print(cambio_genero.tail(1))
```

```python
plt.figure(figsize=(12, 7))

cambio_genero.sort_values().plot(
    kind="barh"
)

plt.title("Cambio en ventas por género: 2013 vs 2016")
plt.xlabel("Cambio en ventas globales (millones)")
plt.ylabel("Género")

plt.axvline(0)

plt.show()
```

```python
crecimiento_plataforma = (
    df_ultimos4
    .groupby(["Year", "Platform"])["Global_Sales"]
    .sum()
    .unstack(fill_value=0)
)

crecimiento_plataforma
```

```python
cambio_plataforma = (
    crecimiento_plataforma.loc[2016] -
    crecimiento_plataforma.loc[2013]
).sort_values(ascending=False)

cambio_plataforma
```

```python
print("Plataforma con mayor crecimiento:")
print(cambio_plataforma.head(1))

print("\nPlataforma con menor crecimiento:")
print(cambio_plataforma.tail(1))
```

```python
plt.figure(figsize=(12, 7))

cambio_plataforma.sort_values().plot(
    kind="barh"
)

plt.title("Cambio en ventas por plataforma: 2013 vs 2016")
plt.xlabel("Cambio en ventas globales (millones)")
plt.ylabel("Plataforma")

plt.axvline(0)

plt.show()
```

### Análisis por videojuego

El dataset no contiene ventas anuales repetidas para cada videojuego,
sino principalmente el año de lanzamiento y sus ventas acumuladas.

Por esta razón, no es posible calcular un crecimiento temporal real
por videojuego. En su lugar, se identifican los videojuegos con mayor
y menor nivel de ventas dentro del periodo analizado (2013-2016).

```python
juegos_periodo = df_ultimos4[
    [
        "Name",
        "Year",
        "Platform",
        "Genre",
        "Global_Sales"
    ]
].sort_values(
    "Global_Sales",
    ascending=False
)

juegos_periodo.head(10)
```

```python
print("Videojuego con mayores ventas:")
display(juegos_periodo.head(1))
```

```python
print("Videojuego con menores ventas:")
display(juegos_periodo.tail(1))
```

```python
oportunidad_genero = (
    df_ultimos4
    .groupby("Genre")[
        ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]
    ]
    .sum()
)

oportunidad_genero
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    oportunidad_genero,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)

plt.title("Ventas por género y región (2013-2016)")
plt.xlabel("Región")
plt.ylabel("Género")

plt.show()
```

```python
oportunidad_promedio = (
    df_ultimos4
    .groupby("Genre")
    .agg(
        Numero_Juegos=("Name", "count"),
        Ventas_Totales=("Global_Sales", "sum"),
        Venta_Promedio=("Global_Sales", "mean")
    )
    .sort_values(
        "Venta_Promedio",
        ascending=False
    )
)

oportunidad_promedio
```

```python
plt.figure(figsize=(12, 7))

oportunidad_promedio[
    "Venta_Promedio"
].sort_values().plot(
    kind="barh"
)

plt.title("Venta global promedio por videojuego y género")
plt.xlabel("Venta promedio (millones)")
plt.ylabel("Género")

plt.show()
```

### Insight de oportunidad 1: Preferencias regionales

Las preferencias de género presentan diferencias importantes entre regiones.
Action y Shooter concentran una parte importante de las ventas en Norteamérica
y Europa, mientras que Role-Playing destaca especialmente en Japón.

Esto sugiere que las oportunidades de mercado pueden variar dependiendo de
la región y que una estrategia de producto debería considerar las preferencias
locales de los consumidores.

### Insight de oportunidad 2: Rendimiento promedio por género

Shooter presenta la mayor venta promedio por videojuego durante el periodo
2013-2016, aunque Action registra un mayor volumen total de ventas.

Esto indica que un género con mayor volumen total no necesariamente presenta
el mayor rendimiento promedio por lanzamiento. El análisis conjunto de ambas
métricas permite identificar oportunidades con mayor precisión.
